Exploratory Data Analysis Findings

1. Dataset Overview

The UNSW-NB15 dataset contains summarized network-flow records classified as normal or malicious traffic.

Dataset	Rows	Columns
Training set	175,341	45
Testing set	82,332	45

Each dataset contains:

39 numerical input features
3 categorical input features: proto, service, and state
1 identifier column: id
1 binary target: label
1 attack-category target: attack_cat

After removing id, label, and attack_cat, 42 input features remain for model training. The testing dataset will remain untouched until final evaluation.

2. Data Quality

Missing values: 0
Infinite numerical values: 0
Every training column contains 175,341 non-null values.
The - value in service represents an unidentified service and is treated as a valid category.
Repeated flow records will be retained for the initial baseline because identical summarized network flows may be legitimate observations.
Extreme numerical values will not be removed automatically because unusual traffic can represent genuine attacks.

3. Binary Target Distribution

Label	Meaning	Percentage
0	Normal traffic	31.94%
1	Attack traffic	68.06%

Attack traffic represents approximately two-thirds of the training dataset. Therefore, the binary target is moderately imbalanced.

The initial baseline will be trained without resampling. It will be evaluated using accuracy, precision, recall, F1-score, ROC-AUC, and a confusion matrix.

Attack recall is especially important because a false negative represents malicious traffic incorrectly classified as normal.

4. Attack-Category Distribution

Category	Percentage
Normal	31.9378%
Generic	22.8127%
Exploits	19.0446%
Fuzzers	10.3706%
DoS	6.9944%
Reconnaissance	5.9832%
Analysis	1.1406%
Backdoor	0.9958%
Shellcode	0.6462%
Worms	0.0741%

The attack-category target is highly imbalanced. Generic, Exploits, and Fuzzers are relatively common, while Analysis, Backdoor, Shellcode, and especially Worms are rare.

Accuracy alone will not reliably evaluate the attack-category classifier. Macro F1-score, weighted F1-score, per-class recall, and a confusion matrix should also be used.

5. Categorical Features

Protocol (proto)
Contains 133 unique categories.
tcp and udp are the most frequent protocols.
unas, arp, and ospf are also relatively common.
Several protocols occur very rarely.
The large number of protocol categories will increase the number of features created by one-hot encoding.
Service (service)
Contains 13 unique categories.
- is the most frequent value.
dns and http are the next most frequent services.
radius, irc, ssl, snmp, and dhcp are rare services.
The - value will be retained as a valid category.
Connection State (state)
Contains 9 unique categories.
INT, FIN, and CON are the most common connection states.
PAR, URN, and no occur only once.
Rare connection states will initially be retained.

The categorical features will be one-hot encoded. Unknown categories must be handled safely when processing validation, testing, or future real-world data.

6. Numerical Features

The numerical features describe different properties of network flows, including:

Flow duration
Source and destination packet counts
Source and destination byte counts
Traffic rate
Time-to-live values
Packet load and packet loss
Packet timing and jitter
TCP window and sequence information
HTTP and FTP activity
Connection-frequency statistics

These numerical features have significantly different value ranges. Therefore, they will be standardized when training Logistic Regression. Tree-based models generally do not require feature scaling.

7. Normal and Attack Traffic Comparison
Feature	Normal mean	Attack mean	Observation
dur	1.0172	1.5200	Attacks have a longer average duration
sbytes	4,105.70	11,068.66	Attacks have approximately 2.7 times more source bytes
dbytes	31,049.46	7,364.46	Normal flows have considerably more destination bytes
spkts	30.73	15.41	Normal flows contain more source packets
dpkts	38.06	10.01	Normal flows contain more destination packets
rate	13,799.31	133,699.69	Attacks have approximately 9.7 times the average traffic rate

These differences suggest that the features may help distinguish normal and malicious traffic. However, differences between group averages do not prove that a feature is predictive.

Extreme values can strongly affect averages, so feature distributions and median values should also be considered.

8. Leakage and Unusable Columns
Binary Intrusion-Detection Model

The binary model predicts whether a network flow is normal or malicious.

Target: label
Remove id because it is only a row identifier.
Remove label because it is the prediction target.
Remove attack_cat because it directly reveals whether the traffic is normal or malicious.
Attack-Category Model

The second model predicts the specific type of network attack.

Target: attack_cat
Use only records where label is equal to 1.
Remove id, label, and attack_cat from the model inputs.
Removing label prevents target information from leaking into the classifier.
9. Preprocessing Decisions

The preprocessing pipeline will:

Remove identifiers, target columns, and leakage columns.
Retain repeated network-flow records for the initial baseline.
Retain unusual numerical values unless they are proven invalid.
Retain - as a valid service category.
One-hot encode proto, service, and state.
Scale numerical features for Logistic Regression.
Safely handle previously unseen categorical values.
Fit preprocessing transformations using training data only.
Use a stratified training-validation split.
Keep the official testing dataset untouched until final evaluation.
10. Conclusion

The UNSW-NB15 dataset is suitable for developing the NetGuard-ML intrusion-detection system.

The binary target is moderately imbalanced, with attacks representing 68.06% of the training records. The attack-category target is highly imbalanced, with several rare attack types.

Categorical features require encoding, while numerical features require scaling for Logistic Regression. The id, label, and attack_cat columns must be handled carefully to prevent data leakage.

The next step is to create a reusable preprocessing pipeline and train the binary Logistic Regression baseline.
